```python
#@title ⚙️ 2. Chạy nội suy video (Tạo Montage Side-by-Side)
MODEL_TYPE = "modify" #@param ["original", "modify"]
ACTIVATION = "smooth_prelu" #@param ["prelu", "smooth_prelu", "gelu", "silu", "soft_clamp_relu"]
ATTENTION = "eca" #@param ["none", "eca"]
INPUT_VIDEO = "demo/demo.mp4" #@param {type:"string"}
OUTPUT_VIDEO = "demo/montage_output.mp4" #@param {type:"string"}

MODEL_DIR = f"trained_model/modify_{ATTENTION}_{ACTIVATION}" if MODEL_TYPE == "modify" else "trained_model/baseline_prelu"

!python inference_video.py \
    --model_type {MODEL_TYPE} \
    --act {ACTIVATION} \
    --attn {ATTENTION} \
    --model {MODEL_DIR} \
    --video {INPUT_VIDEO} \
    --montage \
    --output {OUTPUT_VIDEO}
```

```python
# 3. PHÁT VIDEO TRỰC TIẾP TRÊN COLAB
import base64
import io
from IPython.display import HTML

if os.path.exists(OUTPUT_VIDEO):
    video = io.open(OUTPUT_VIDEO, 'r+b').read()
    encoded = base64.b64encode(video)
    HTML(data='''<video alt="RIFE Output" controls style="max-width:100%; border-radius: 8px;">
                    <source src="data:video/mp4;base64,{0}" type="video/mp4" />
                 </video>'''.format(encoded.decode('ascii')))
else:
    print(f"⚠️ Chưa tìm thấy file video {OUTPUT_VIDEO}. Hãy chạy bước 2 để xuất video.")
```
